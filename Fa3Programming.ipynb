```python
#fa3prog2
Score = int(input("Enter Your Exam Score!"))

if Score > 100:
  print("That's not a valid input!")
elif Score < 75:
  print("You Have Failed! Try Your Best Next Time!")
elif Score >= 75:
  print("You Have Passed! Congratulations!")
else:
  print("Invalid Input!")
```
